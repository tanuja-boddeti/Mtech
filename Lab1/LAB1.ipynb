```python
# 1 Given a string count the number of vowels and return the count
def vowel(s):
  count=0
  vowels="aeiouAEIOU"
  for char in s:
    if char in vowels:
      count+=1
  return count
s=input("Enter a string:")
print(vowel(s))
```

```python
# 2 Give a string count the number of each vowel in the string and return it as a dictionary where key is a vowel and value is count of that respective vowel.
def count_each_vowel(s):
    vowels = 'aeiou'
    vowel_count = {v: 0 for v in vowels}
    for char in s.lower():
        if char in vowel_count:
            vowel_count[char] += 1
    return vowel_count
s=input("Enter a string:")
print(count_each_vowel(s))
```

```python
# 3 Write a function to check if a number is prime or not.
def prime(num):
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
number = int(input("Enter a number: "))
if prime(number):
    print(number, "is a prime number")
else:
    print(number, "is not a prime number")
```

```python
# 4 Write a Python program to print even-length words in a string.
def even_length_words(string):
    words = string.split(" ")
    print("Even-length words are:")
    for word in words:
        if len(word) % 2 == 0:
            print(word)
string = input("Enter a string:")
even_length_words(string)
```

```python
# 5 Write a Python function to print: 1 2 2 3 3 3
def print_pattern(n):
    for i in range(1, n+1):
        for j in range(i):
            print(i, end=" ")
        print()
n=int(input("enter a number:"))
print_pattern(n)
```

```python
# 6 Write a Python program to check whether a string is symmetrical or palindrome.
def check_string(s):
    if s == s[::-1]:
        print("The string is a palindrome")
    else:
        print("The string is not a palindrome")
    n = len(s)
    if n % 2 == 0:
        if s[:n//2] == s[n//2:]:
            print("The string is symmetrical")
        else:
            print("The string is not symmetrical")
    else:
        print("The string is not symmetrical")
string = input("Enter a string: ")
check_string(string)
```

```python
""" 9 Create a class named as “operations” where each method is an arithmetic operation
(Addition, subtraction, multiplication, division). (Note each method should get an
argument and within the constructor get at least one argument)"""

class operations:
    def __init__(self, num):
        self.num = num
    def addition(self, x):
        return self.num + x
    def subtraction(self, x):
        return self.num - x
    def multiplication(self, x):
        return self.num * x
    def division(self, x):
        return self.num / x

obj = operations(10)

print("Addition:", obj.addition(5))
print("Subtraction:", obj.subtraction(5))
print("Multiplication:", obj.multiplication(5))
print("Division:", obj.division(5))
```

```python
# 10 Given a list of numbers, sort it into two different lists of even and odd numbers separately.
def sort(l):
  even_list=[]
  odd_list=[]
  for i in l:
    if i%2==0:
      even_list.append(i)
    else:
      odd_list.append(i)
  return even_list,odd_list

l= list(map(int, input("Enter list of numbers: ").split()))
print(sort(l))
```

```python
""" 11 Given a dictionary of marks
marks = {
"math": 80,
"english": 75,
"science": 90,
"computer": 95
}
Return the subject name that has maximum marks"""

def maximum_marks(marks):
    max_marks = 0
    max_subject = ""

    for subject in marks:
        if marks[subject] > max_marks:
            max_marks = marks[subject]
            max_subject = subject

    return max_subject
marks = {
    "math": 80,
    "english": 75,
    "science": 90,
    "computer": 95
}

print("Subject with maximum marks:", maximum_marks(marks))
```

```python
#12. Given a list of numbers with duplicates, return True if there is at least one duplicate in the list.
def check_duplicate(numbers):
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] == numbers[j]:
                return True
    return False
numbers = list(map(int, input("Enter numbers: ").split()))
print(check_duplicate(numbers))
```

```python
# 14. Best Time to Buy and Sell Stock – Leetcode problem
def max_profit(prices):
    min_price = prices[0]
    profit = 0
    for price in prices:
        if price < min_price:
            min_price = price
        current_profit = price - min_price
        if current_profit > profit:
            profit = current_profit
    return profit


prices = list(map(int, input("Enter stock prices: ").split()))

print(max_profit(prices))
```

```python
# 15 Given a list of numbers nums = [100,4,200,1,3,2] find the longest consecutive sequence. For the given list the answer is 4 because the longest consecutive sequence is 1,2,3,4

def longest_sequence(nums):
    nums.sort()

    longest = 1
    current = 1

    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1] + 1:
            current += 1
        elif nums[i] != nums[i - 1]:
            current = 1

        if current > longest:
            longest = current

    return longest


l= list(map(int, input("Enter list of numbers: ").split()))

print("Longest consecutive sequence:", longest_sequence(l))
```
